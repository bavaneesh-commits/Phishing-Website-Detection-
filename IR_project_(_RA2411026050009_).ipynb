```python
# Step 4: Data cleaning and feature preparation

import pandas as pd

# 1. Combine features and target for inspection
df = pd.concat([X, y], axis=1)

print("Original dataset shape:", df.shape)

# 2. Check missing values
print("\nMissing values in each column:")
print(df.isnull().sum())

print("\nTotal missing values:", df.isnull().sum().sum())

# 3. Check duplicate records
print("\nNumber of completely duplicated records:", df.duplicated().sum())

# 4. Check duplicate feature rows
print("Number of duplicate feature rows:", X.duplicated().sum())

# 5. Check target labels
target_column = y.columns[0]

print("\nTarget column:", target_column)
print("Target labels:", y[target_column].unique())
print("\nTarget label counts:")
print(y[target_column].value_counts(dropna=False))

# 6. Verify whether all features are numeric
print("\nFeature data types:")
print(X.dtypes)

non_numeric_columns = X.select_dtypes(
    exclude="number"
).columns.tolist()

print("\nNon-numeric feature columns:", non_numeric_columns)

# 7. Create clean working copies
X_clean = X.copy()
y_clean = y[target_column].copy()

# 8. Stop if missing values require a deliberate handling decision
if X_clean.isnull().any().any() or y_clean.isnull().any():
    print("\nWARNING: Missing values exist.")
    print("We must decide how to handle them before model training.")
else:
    print("\nNo missing values found in features or target.")

# 9. Check whether the target contains exactly two classes
print("\nNumber of target classes:", y_clean.nunique())

if y_clean.nunique() == 2:
    print("The target contains two classes.")
else:
    print("The target does not contain exactly two classes.")
```

```python
# Step 5: Split the dataset into training and testing sets

from sklearn.model_selection import train_test_split

# Separate features and target
X = X_clean.copy()
y = y_clean.copy()

# Check data before splitting
print("Feature dataset shape:", X.shape)
print("Target dataset shape:", y.shape)

print("\nTarget class distribution:")
print(y.value_counts())

# Split the dataset: 80% training and 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Display the resulting dimensions
print("\nTraining and testing dataset dimensions")
print("-" * 40)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)

# Verify class proportions
print("\nTraining class distribution:")
print(y_train.value_counts(normalize=True).round(4) * 100)

print("\nTesting class distribution:")
print(y_test.value_counts(normalize=True).round(4) * 100)

print("\nDataset splitting completed successfully.")
```

```python
# Step 6: Inspect numerical features and prepare preprocessing

from sklearn.preprocessing import StandardScaler

# 1. Check whether all features are numerical
print("All features are numerical:",
      all(X_train.dtypes.apply(lambda dtype: dtype.kind in "iuf")))

# 2. Inspect the actual values in the first five features
print("\nFirst five feature columns:")
print(X_train.columns[:5].tolist())

print("\nSample feature values:")
display(X_train.iloc[:5, :5])

# 3. Count unique values in the feature columns
print("\nNumber of unique values per feature:")
display(X_train.nunique().sort_values().to_frame("Unique Values"))

# 4. Check the overall range of feature values
print("\nMinimum feature values:")
display(X_train.min())

print("\nMaximum feature values:")
display(X_train.max())

# 5. Create the scaler
scaler = StandardScaler()

print("\nStandardScaler created successfully.")
print("It has NOT been fitted yet.")
print("We will fit it only on the training data inside the model pipelines.")
```

```python
# Step 7: Prepare four machine learning classification models

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB

# 1. Decision Tree
decision_tree = Pipeline([
    ("classifier", DecisionTreeClassifier(random_state=42))
])

# 2. K-Nearest Neighbors
knn = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", KNeighborsClassifier(n_neighbors=5))
])

# 3. Support Vector Machine
svm = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", SVC(kernel="rbf", C=1.0, random_state=42))
])

# 4. Gaussian Naive Bayes
naive_bayes = Pipeline([
    ("classifier", GaussianNB())
])

# Store the models in a dictionary
models = {
    "Decision Tree": decision_tree,
    "KNN": knn,
    "SVM": svm,
    "Gaussian Naive Bayes": naive_bayes
}

# Display model configuration
print("Four classification models prepared successfully.")
print("-" * 45)

for name, model in models.items():
    print("\nModel:", name)
    print(model)

print("\nTraining records:", len(X_train))
print("Testing records:", len(X_test))
print("\nThe models have NOT been trained yet.")
```

```python
# Step 8: Train and evaluate four machine learning models

import time
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

results = []

for name, model in models.items():

    print(f"\nTraining {name}...")

    # Measure training time
    start_train = time.perf_counter()

    model.fit(X_train, y_train)

    end_train = time.perf_counter()

    training_time = end_train - start_train

    # Measure prediction time
    start_predict = time.perf_counter()

    y_pred = model.predict(X_test)

    end_predict = time.perf_counter()

    prediction_time = end_predict - start_predict

    # Calculate evaluation metrics
    accuracy = accuracy_score(y_test, y_pred)

    precision = precision_score(
        y_test, y_pred, average="binary", zero_division=0
    )

    recall = recall_score(
        y_test, y_pred, average="binary", zero_division=0
    )

    f1 = f1_score(
        y_test, y_pred, average="binary", zero_division=0
    )

    # Store the actual results
    results.append({
        "Model": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1,
        "Training Time (seconds)": training_time,
        "Prediction Time (seconds)": prediction_time
    })

    print(f"{name} training completed.")

# Create a comparison DataFrame
results_df = pd.DataFrame(results)

print("\nMODEL PERFORMANCE COMPARISON")
display(results_df.round(4))
```

```python
# Step 9: Generate confusion matrices for all four models

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import confusion_matrix, classification_report

# Check the actual target labels
print("Actual target labels:", sorted(y.unique()))

# Create a confusion matrix for each model
for name, model in models.items():

    # Predict test data
    y_pred = model.predict(X_test)

    # Calculate confusion matrix
    cm = confusion_matrix(y_test, y_pred)

    print("\n" + "=" * 50)
    print("MODEL:", name)
    print("=" * 50)

    print("\nConfusion Matrix:")
    print(cm)

    print("\nClassification Report:")
    print(
        classification_report(
            y_test,
            y_pred,
            zero_division=0
        )
    )

    # Plot confusion matrix
    plt.figure(figsize=(6, 4))

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues"
    )

    plt.title(f"Confusion Matrix - {name}")
    plt.xlabel("Predicted Label")
    plt.ylabel("Actual Label")
    plt.tight_layout()
    plt.show()
```

```python
# Step 10: Visual comparison of the four models

import matplotlib.pyplot as plt
import seaborn as sns

# Display the actual comparison table
print("MODEL COMPARISON TABLE")
display(results_df.round(4))

# Select evaluation metrics
metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1-Score"
]

# Convert results into a format suitable for plotting
plot_data = results_df[["Model"] + metrics].melt(
    id_vars="Model",
    var_name="Metric",
    value_name="Score"
)

# Create comparison plot
plt.figure(figsize=(12, 6))

sns.barplot(
    data=plot_data,
    x="Model",
    y="Score",
    hue="Metric"
)

plt.title("Performance Comparison of Four Classification Algorithms")
plt.xlabel("Machine Learning Algorithm")
plt.ylabel("Evaluation Score")
plt.ylim(0, 1.05)
plt.xticks(rotation=15)
plt.legend(title="Evaluation Metrics")
plt.tight_layout()
plt.show()

# Identify the model with the highest F1-score
best_f1_model = results_df.loc[
    results_df["F1-Score"].idxmax()
]

print("\nMODEL WITH THE HIGHEST F1-SCORE")
print("Model:", best_f1_model["Model"])
print("F1-score:", round(best_f1_model["F1-Score"], 4))
print("Accuracy:", round(best_f1_model["Accuracy"], 4))
print("Precision:", round(best_f1_model["Precision"], 4))
print("Recall:", round(best_f1_model["Recall"], 4))
```

```python
# Step 11: SVM what-if experiment

import time
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

# Values of C to compare
C_values = [0.1, 1, 10]

svm_results = []

for c_value in C_values:

    print(f"\nTraining SVM with C = {c_value}")

    # Create a fresh pipeline for each experiment
    svm_model = Pipeline([
        ("scaler", StandardScaler()),
        ("classifier", SVC(
            kernel="rbf",
            C=c_value,
            random_state=42
        ))
    ])

    # Measure training time
    start_train = time.perf_counter()

    svm_model.fit(X_train, y_train)

    training_time = time.perf_counter() - start_train

    # Measure prediction time
    start_predict = time.perf_counter()

    y_pred = svm_model.predict(X_test)

    prediction_time = time.perf_counter() - start_predict

    # Calculate actual evaluation metrics
    svm_results.append({
        "C": c_value,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(
            y_test, y_pred,
            average="binary",
            zero_division=0
        ),
        "Recall": recall_score(
            y_test, y_pred,
            average="binary",
            zero_division=0
        ),
        "F1-Score": f1_score(
            y_test, y_pred,
            average="binary",
            zero_division=0
        ),
        "Training Time (seconds)": training_time,
        "Prediction Time (seconds)": prediction_time
    })

# Create the results table
svm_results_df = pd.DataFrame(svm_results)

print("\nSVM WHAT-IF EXPERIMENT RESULTS")
display(svm_results_df.round(4))

# Identify the best C value by F1-score
best_svm_result = svm_results_df.loc[
    svm_results_df["F1-Score"].idxmax()
]

print("\nBest C value based on F1-score:")
print("C =", best_svm_result["C"])
print("F1-score =", round(best_svm_result["F1-Score"], 4))
```
